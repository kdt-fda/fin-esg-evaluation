```python
# DB 생성
with conn.cursor(pymysql.cursors.DictCursor) as cur:
    cur.execute('CREATE DATABASE STOCK_DB CHARACTER SET utf8mb4 COLLATE utf8mb4_unicode_ci;')
    cur.execute('USE STOCK_DB;')
```

```python
# DATE_TB 테이블 생성
with conn.cursor(pymysql.cursors.DictCursor) as cur:
    cur.execute("""
                CREATE TABLE DATE_TB (
                    trade_date DATE NOT NULL,
                    PRIMARY KEY(trade_date)
                );
                """)
```

```python
# KOSPI200_STOCKS_TB 테이블 생성
with conn.cursor(pymysql.cursors.DictCursor) as cur:
    cur.execute("""
                CREATE TABLE KOSPI200_STOCKS_TB (
                    ticker VARCHAR(10) NOT NULL,
                    stock_name VARCHAR(100),
                    PRIMARY KEY(ticker)
                );
                """)
```

```python
# SECTOR_TB 테이블 생성
with conn.cursor(pymysql.cursors.DictCursor) as cur:
    cur.execute("""
                CREATE TABLE SECTOR_TB (
                    sector_code VARCHAR(10) NOT NULL,
                    sector_name VARCHAR(100) NOT NULL,
                    PRIMARY KEY(sector_code)
                );
                """)
```

```python
# STOCK_SECTOR_TB 테이블 생성
with conn.cursor(pymysql.cursors.DictCursor) as cur:
    cur.execute("""
                CREATE TABLE STOCK_SECTOR_TB (
                    ticker VARCHAR(10) NOT NULL,
                    sector_code VARCHAR(10) NOT NULL,
                    PRIMARY KEY(ticker),
                    CONSTRAINT fk_ss_ticker
                        FOREIGN KEY(ticker) REFERENCES KOSPI200_STOCKS_TB(ticker),
                    CONSTRAINT fk_ss_sector
                        FOREIGN KEY(sector_code) REFERENCES SECTOR_TB(sector_code)
                );
                """)
```

```python
# STOCK_TB 테이블 생성
with conn.cursor(pymysql.cursors.DictCursor) as cur:
    cur.execute("""
                CREATE TABLE STOCK_TB (
                    trade_date DATE NOT NULL,
                    ticker VARCHAR(10) NOT NULL,
                    stock_name VARCHAR(100),
                    open INT,
                    high INT,
                    low INT,
                    close INT,
                    volume BIGINT,
                    foreign_net_amt BIGINT,
                    inst_net_amt BIGINT,
                    ma5 DECIMAL(16,6),
                    ma20 DECIMAL(16,6),
                    ma60 DECIMAL(16,6),
                    ma120 DECIMAL(16,6),
                    bb_upper DECIMAL(16,6),
                    bb_lower DECIMAL(16,6),
                    bb_breakout TINYINT(1),
                    msci_event BOOLEAN,
                    rsi DECIMAL(16,6),
                    macd DECIMAL(16,6),
                    macd_signal DECIMAL(16,6),
                    golden_cross_5_20 BOOLEAN,
                    death_cross_5_20 BOOLEAN,
                    golden_cross_20_60 BOOLEAN,
                    death_cross_20_60 BOOLEAN,
                    PRIMARY KEY(trade_date, ticker),
                    CONSTRAINT fk_stock_date
                        FOREIGN KEY(trade_date) REFERENCES DATE_TB(trade_date),
                    CONSTRAINT fk_stock_ticker
                        FOREIGN KEY(ticker) REFERENCES KOSPI200_STOCKS_TB(ticker)
                );
                """)
```
